`````python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-0"
`````

`````python
def add_user_message(messages, content):
    messages.append({"role": "user", "content": content})
    return messages

def add_assistant_message(messages, content):
    messages.append({"role": "assistant", "content": content})
    return messages

def chat(messages, system_prompt=None, temperature=0.5, stop_sequences=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature
    }
    if stop_sequences:
        params["stop_sequences"] = stop_sequences
    if system_prompt:
        params["system"] = system_prompt
    message = client.messages.create(**params)
    return message.content[0].text
`````

`````python
message = []
message = add_user_message(message, "Define quantum computing in one sentence.")
answer = chat(message)
answer
`````

`````python
message = add_assistant_message(message, answer)
message = add_user_message(message, "Write another sentence")
answer = chat(message)
answer
`````

`````python
message = []

print("Bienvenido al primer chat")
while True:
    user_input = input(">")
    print(">>>>>", user_input)
    message = add_user_message(message, user_input)
    answer = chat(message)
    print("<<<", answer)
    message = add_assistant_message(message, answer)
    print("<", answer)
`````
